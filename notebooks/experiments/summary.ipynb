```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../logs/evaluation.csv")
```

```python
# Разделяем эксперименты
df_without_aug = df.iloc[:6].copy()
df_with_aug = df.iloc[6:].copy()
```

```python
# Строим график
plt.figure(figsize=(10, 6))

plt.plot(
    df_without_aug["epochs"],
    df_without_aug["accuracy"],
    marker="o",
    label="Без аугментации"
)

plt.plot(
    df_with_aug["epochs"],
    df_with_aug["accuracy"],
    marker="o",
    label="С аугментацией"
)

plt.xlabel("Количество эпох")
plt.ylabel("Accuracy")
plt.title("Влияние аугментации на качество модели")

plt.legend()
plt.grid(True)

plt.show()
```

Вывод: аугментация помогла модели запомнить более устойчивые признаки для классификации изображений и повысить метрику качества
